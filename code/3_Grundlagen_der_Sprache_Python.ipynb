### 3.12 Beispielprogramm: Wörter zählen

```python
text = input("Bitte einen Text eingeben: ")
# Zerlege den Text in Wörter anhand von Leerzeichen
woerter = text.split()
# Anzahl ermitteln
anzahl = len(woerter)
print(f"Der Text enthält {anzahl} Wörter.")
```

```python
text = input("Bitte einen Text eingeben: ")
# Wörter, die nicht gezählt werden sollen
stoppwoerter = ["der", "die", "das", "ein", "eine", "und", "oder", 
                "von", "zu", "mit"]
# Text in Wörter aufteilen
woerter = text.split()
# Wörter zählen, die nicht in der Stoppwortliste sind
anzahl = 0
for wort in woerter:
    # Vergleich in Kleinbuchst., um Groß-/Kleinschreibung zu ignorieren
    if wort.lower() not in stoppwoerter:
        anzahl += 1

print(f"Der Text enthält {anzahl} Wörter (ohne Stoppwörter).")
```

```python
text = input("Bitte einen Text eingeben: ")

# Text in Wörter aufteilen
woerter = text.split()

# Wörter zählen
haeufigkeit = {}
for wort in woerter:
    wort = wort.lower()  # Groß-/Kleinschreibung ignorieren
    if wort in haeufigkeit:
        haeufigkeit[wort] += 1
    else:
        haeufigkeit[wort] = 1

# Wort mit höchster Häufigkeit finden
max_wort = max(haeufigkeit, key=haeufigkeit.get)
max_anzahl = haeufigkeit[max_wort]

print(f"Häufigstes Wort: '{max_wort}' ({max_anzahl} mal)")
```

### 3.13 Eigene Funktionen schreiben

```python
def wortanzahl(text):
    woerter = text.split()
    return len(woerter)

# Beispiel
satz = "Das ist ein kurzer Beispielsatz."
print(wortanzahl(satz))  # Ausgabe: 5
```

```python
def addieren(x, y):
    ergebnis = x + y
    return ergebnis
z = addieren(5, 3)
print(z)            # Hier wird 8 ausgegeben.
print(ergebnis)     # Fehler!
```

```python
def addieren(x, y=1):      # y ist ein optionaler Parameter.
   return x + y
print(addieren(5, 3))      # 5+3 = 8
print(addieren(5))         # 5+1 = 6
```

```python
def wortanzahl(text, min_len=1):
    woerter = text.split()
    gefiltert = [wort for wort in woerter if len(wort) >= min_len]
    return len(gefiltert)
satz = "Das ist ein kurzer Satz."
print(wortanzahl(satz))              # Ausgabe: 5
print(wortanzahl(satz, min_len=4))   # Ausgabe: 2 ("kurzer", "Satz.")
```

### 3.14 Pythonischer Code

```python
woerter = "Das ist ein kurzer Satz.".split()
min_len = 4

gefiltert = []
for wort in woerter:
    if len(wort) >= min_len:
        gefiltert.append(wort)

print(gefiltert)

# List Comprehension:
gefiltert = [wort for wort in woerter if len(wort) >= min_len]      # pythonisch 😎

print(gefiltert)
```

**Walrus-Operator**

```python
anrede = "Herr"

if anrede.upper() == "HERR":
    print(f"HALLO {anrede.upper()} {nachname}")  # unpythonisch

if (anrede2 := anrede.upper()) == "HERR":
    print(f"HALLO {anrede2} {nachname}")         # pythonisch 😎
```
